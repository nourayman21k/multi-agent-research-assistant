```python
Model_Cache ={}
```

```python
import time
def cache_model(query):
    start_time=time.time()
    if Model_Cache.get(query):
        print("**CAche Hit**")
        end_time=time.time()
        elapsed_time=end_time-start_time
        print(f"EXECUTION TIME: {elapsed_time:.2f} seconds")
        return Model_Cache.get(query)
    else:
        print("***CACHE MISS – EXECUTING MODEL***")
        start_time = time.time()
        response = llm.invoke(query)
        end_time = time.time()
        elapsed = end_time - start_time
        print(f"EXECUTION TIME: {elapsed:.2f} seconds")
        Model_Cache[query] = response
        return response
```

### advanced CAG

```python
from __future__ import annotations
from typing import TypedDict, List, Optional
import time

# ---- LangGraph / LangChain ----
from langgraph.graph import StateGraph, END
from langgraph.checkpoint.memory import MemorySaver

from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
# ---- FAISS vector stores ----
import faiss
from langchain_community.vectorstores import FAISS
from langchain_community.docstore.in_memory import InMemoryDocstore
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

EMBED = HuggingFaceEmbeddings(
    model_name="dmis-lab/UltraBERT-Embed-150M",
    model_kwargs={"trust_remote_code": True},
    encode_kwargs={"normalize_embeddings": True},
)
VECTOR_DIM = 384
```

```python
from pydantic import BaseModel, Field
class RAG_State(BaseModel):
    question: str = ""
    normalized_question: str = ""
    cache_hit: bool = False
    context_docs: List = Field(default_factory=list)
    answer: Optional[str] = None
    citations: List[str] = Field(default_factory=list)
```

```python
# ----- QA CACHE (EMPTY, SAFE INIT) -----
VECTOR_DIM = 384
qa_index = faiss.IndexFlatL2(VECTOR_DIM)  # distance; lower is better
QA_CACHE = FAISS(
    embedding_function=EMBED,
    index=qa_index,
    docstore=InMemoryDocstore({}),
    index_to_docstore_id={}
)
```

```python
QA_CACHE
```

```python
# ----- RAG STORE (demo only) -----
RAG_STORE = FAISS.from_texts(
    texts=[
        "LangGraph lets you compose stateful LLM workflows as graphs.",
        "In LangGraph, nodes can be cached; node caching memoizes outputs keyed by inputs for a TTL.",
        "Retrieval-Augmented Generation (RAG) retrieves external context and injects it into prompts.",
        "Semantic caching reuses prior answers when new questions are semantically similar."
    ],
    embedding=EMBED
)
```

```python
llm
```

```python

CACHE_TOP_K = 3

CACHE_DISTANCE_THRESHOLD = 0.45
RETRIEVE_TOP_K = 4
# Optional TTL for cache entries (seconds). 0 = disabled.
CACHE_TTL_SEC = 0
```

```python
import re
def normalize_query(state: RAG_State) -> RAG_State:
    q = " ".join((state.question or "").split())   # trim + collapse whitespace
    state.normalized_question = q.lower()
    return state

def semantic_cache_lookup(state: RAG_State) -> RAG_State:
    q = state.normalized_question
    state.cache_hit = False

    if not q:
        return state
    if getattr(QA_CACHE, "index", None) is None or QA_CACHE.index.ntotal == 0:
        return state

    hits = QA_CACHE.similarity_search_with_score(q, k=CACHE_TOP_K)

    for doc, dist in hits:                      # best-first; take first valid hit
        if dist > CACHE_DISTANCE_THRESHOLD:
            break                               # sorted, so the rest are worse
        if CACHE_TTL_SEC > 0:
            ts = doc.metadata.get("ts")
            if ts is None or (time.time() - float(ts)) > CACHE_TTL_SEC:
                continue
        answer = doc.metadata.get("answer")
        if answer:
            state.answer = answer
            state.citations = ["(cache)"]
            state.cache_hit = True
            break
    return state

def respond_from_cache(state: RAG_State) -> RAG_State:
    return state

def retrieve(state: RAG_State) -> RAG_State:
    state.context_docs = RAG_STORE.similarity_search(
        state.normalized_question, k=RETRIEVE_TOP_K
    )
    return state

def generate(state: RAG_State) -> RAG_State:
    question = state.normalized_question
    docs = state.context_docs
    ctx = "\n\n".join(f"[doc-{i}] {d.page_content}" for i, d in enumerate(docs, 1))

    system = (
        "You are a precise RAG assistant. Use the context when helpful. "
        "Cite with [doc-i] markers if you use a fact from the context."
    )
    user = f"Question: {question}\n\nContext:\n{ctx}\n\nWrite a concise answer with citations."

    resp = llm.invoke([
        {"role": "system", "content": system},
        {"role": "user", "content": user},
    ])
    state.answer = resp.content

    used = sorted({int(i) for i in re.findall(r"\[doc-(\d+)\]", state.answer)})
    state.citations = [f"[doc-{i}]" for i in used if i <= len(docs)]
    return state

def cache_write(state: RAG_State) -> RAG_State:
    if state.cache_hit or not state.normalized_question or not state.answer:
        return state
    QA_CACHE.add_texts(
        texts=[state.normalized_question],
        metadatas=[{"answer": state.answer, "ts": time.time()}],
    )
    return state
```

```python
from langgraph.graph import StateGraph, END

g = StateGraph(RAG_State)
g.add_node("normalize_query", normalize_query)
g.add_node("semantic_cache_lookup", semantic_cache_lookup)
g.add_node("respond_from_cache", respond_from_cache)
g.add_node("retrieve", retrieve)
g.add_node("generate", generate)
g.add_node("cache_write", cache_write)

g.set_entry_point("normalize_query")
g.add_edge("normalize_query", "semantic_cache_lookup")
g.add_conditional_edges(
    "semantic_cache_lookup",
    lambda s: "hit" if s.cache_hit else "miss",
    {"hit": "respond_from_cache", "miss": "retrieve"},
)
g.add_edge("respond_from_cache", END)
g.add_edge("retrieve", "generate")
g.add_edge("generate", "cache_write")
g.add_edge("cache_write", END)

app = g.compile()
result = app.invoke(RAG_State(question="What is FAISS?"))
```
